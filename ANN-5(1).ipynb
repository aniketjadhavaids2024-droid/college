##### Title : Write a python Program for Bidirectional Associative Memory with two pairs of vectors.

```python
import numpy as np

x1 = np.array([1,1,-1,-1])
y1 = np.array([1,-1,1])

x2 = np.array([1,-1,1,1])
y2 = np.array([-1,1,-1])

print("x1 =", x1)
print("y1 =", y1)

print("x2 =", x2)
print("y2 =", y2)

```

```python
W = np.outer(x1, y1) + np.outer(x2, y2)

print("Weight Matrix W:")
print(W)
```

```python
def sign_function(X):
    return np.where(X>=0 , 1, -1)
```

```python
def forword(x):
    return sign_function(x@W)
```

```python
def backword(y):
    return sign_function(y@W.T)
```

```python
Y = forword(x1)
print("X1 ->  :",x1)
print("Expected    :", y1)
print("Predicted Y :", Y)
print()
X = backword(y1)
print("y1 ->  :",y1)
print("Expected    :", x1)
print("Predicted X :", X)
```
