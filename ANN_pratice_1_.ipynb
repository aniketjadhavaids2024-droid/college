```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x=np.linspace(-10,10,100)
```

```python
step=np.where(x>=0,1,0)
```

```python
sigmoid=1/(1+np.exp(-x))
print(step)
print("Total:", np.sum(step))
```

```python
tanh=np.tanh(x)
print(tanh)
print("Total:", np.sum(tanh))
```

```python
relu=np.maximum(0,x)
print(relu)
print("Total:", np.sum(relu))
```

```python
leaky_relu=np.maximum(0.01*x,x)
print(leaky_relu)
print("Total:", np.sum(leaky_relu))
```

```python
softmax = np.exp(x) / np.sum(np.exp(x))

print(softmax)
print("Total:", np.sum(softmax))
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(3, 2, 1)
plt.plot(x, step, label="Step")
plt.title("Step Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()
plt.legend()

plt.subplot(3, 2, 2)
plt.plot(x, tanh, label="tanh")
plt.title("tanh Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()
plt.legend()
plt.subplot(3, 2, 3)
plt.plot(x, sigmoid, label="Sigmoid")
plt.title("Sigmoid Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()
plt.legend()

plt.subplot(3, 2, 4)
plt.plot(x, relu, label="ReLU")
plt.title("ReLU Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()
plt.legend()

plt.subplot(3, 2, 5)
plt.plot(x, leaky_relu, label="Leaky ReLU")
plt.title("Leaky ReLU Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()
plt.legend()

plt.tight_layout()
plt.show()
```

```python
functions = [step, tanh, sigmoid, relu, leaky_relu]
names = ["Step", "Tanh", "Sigmoid", "ReLU", "Leaky ReLU"]

plt.figure(figsize=(12, 8))

for i in range(5):
    plt.subplot(3, 2, i+1)
    plt.plot(x, functions[i], label=names[i])
    plt.title(names[i] + " Function")
    plt.xlabel("Input")
    plt.ylabel("Output")
    plt.grid()
    plt.legend()

plt.tight_layout()
plt.show()
```
