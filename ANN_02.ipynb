```python
import numpy as np
inputs=np.array([[0,0],
                 [0,1],
                 [1,0],
                 [1,1]])
```

```python
weights=np.array([1,1])
and_threshold=2
or_threshold=1
```

```python
print("And gate")
print("A B output")
for x in inputs:
    net_input = np.sum(x * weights)
    output = 1 if net_input >= and_threshold else 0
    print(x[0], x[1], output)
```

```python
print("OR gate")
print("A B output")
for x in inputs:
  net_output=np.sum(x*weights)
  output=1 if net_output>=or_threshold else 0
  print(x[0],x[1],output)
```
