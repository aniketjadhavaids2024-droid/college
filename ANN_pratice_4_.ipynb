```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
x=np.array([[2,3],
            [3,3],
            [3,4],
            [1,1],
            [2,1],
            [1,2],
            [1,4]])
```

```python
t=np.array([1,1,1,0,0,0,1])
```

```python
w=np.zeros(2) 
b=-0.5
lr=0.1 
epoches=10
```

```python
for epoch in range(epoches):
    for i in range(len(x)):
        z=np.dot(x[i],w)+b
        y=1 if z>=0 else 0
        error=t[i]-y
        w=w+lr*error*x[i]
        b=b+lr*error 
print("final weight",w)
print("final bias",b)
```

```python

x1=np.linspace(-3,7,100)
x2=np.linspace(0,8,100)
xx,yy=np.meshgrid(x1,x2)
```

```python
z=w[0]*xx+w[1]*yy+b
```

```python
plt.contourf(xx,yy,z>=0,alpha=0.3)
plt.contour(xx,yy,z,levels=[0]) 
plt.scatter(x[:,0],x[:,1],c=t)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Perceptron Decision Regions")
plt.grid()
plt.show()
```
